```python
import collections
```

```python
Card = collections.namedtuple('Card', ['rank', 'suit'])
class FranchDesk:
    def __init__(self):
        ranks = [str(n) for n in range(2,11)] + list('JQKA')
        suits = 'spades diamonds clubs hearts'.split()
        self._cards = [Card(rank,suit) for suit in suits for rank in ranks]
    
    def __len__(self):
        return len(self._cards)
    
    def __getitem__(self, position):
        return self._cards[position]
```

```python
beer_card = Card('7', 'diamonds')
print(beer_card)
```

```python
deck = FranchDesk()
len(deck)
deck[5]
```

```python
from random import choice
choice(deck)
```

```python
Card('Q', 'hearts') in deck
```

迭代往往是隐式的。如果一个容器没有实现 __contains__ 方法,那  么 in 运算符就会做一次顺序扫描。本例就是这样,FrenchDeck 类支  持 in 运算符,因为该类可迭代。下面来试试。

```python
Card('7', 'beasts') in deck
```

```python
deck[12::13]
```

```python
for card in deck:
    print(card)
```

另外,也可以反向迭代这摞纸牌。

```python
for card in reversed(deck):
    print(card)
```

!r: 是 Python 字符串格式化中的转换标志，用于调用对象的 __repr__() 方法。


```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"
    def __str__(self):
        return f"Person: {self.name} is {self.age} years old"

p1 = Person('John', 30)
print(p1)
```

```python
p  = Person("John", 30)
print(f"{p}")
print(f"{p!r}")
```

```python
value = "hello"
print(f"变量值：{value!r}")
```
