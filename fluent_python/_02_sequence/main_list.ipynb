## 列表推导式与 map 和 filter 比较
列表推导式涵盖 map 和 filter 两个函数的功能,写出的代码不像  Python 的 lambda 表达式那样晦涩难懂

````python
symbols = '$¢£¥€¤'
beyond_ascii = [ord(s) for s in symbols if ord(s) > 127]
print(beyond_ascii)
beyond_ascii = list(filter(lambda c: c> 127,map(ord,symbols)))
print(beyond_ascii)
````

## 详细解释：`beyond_ascii = list(filter(lambda c: c> 127,map(ord,symbols)))`

这行代码展示了 Python 中函数式编程的组合使用。让我们从内到外逐步分解：


### 步骤 1：理解 symbols 和 ord() 函数


````python
symbols = '$¢£¥€¤'

# ord() 函数返回字符的 Unicode 码点（整数）
print("字符和对应的 Unicode 码点：")
for char in symbols:
    print(f"'{char}' -> {ord(char)}")

# ASCII 字符的范围是 0-127
# 大于 127 的字符属于扩展字符集（如拉丁字符、中文字符等）
print(f"\n大于 127 的码点（非 ASCII 字符）：")
for char in symbols:
    code = ord(char)
    if code > 127:
        print(f"  '{char}' -> {code}")
````

### 步骤 2：map() 函数 - 将每个字符转换为 Unicode 码点


````python
symbols = '$¢£¥€¤'

# map(ord, symbols) 对 symbols 中的每个字符应用 ord() 函数
# map() 返回一个迭代器（iterator），不是列表
mapped_result = map(ord, symbols)
print(f"map() 返回的类型：{type(mapped_result)}")
print(f"map() 对象：{mapped_result}")

# 转换为列表才能看到所有值
mapped_list = list(map(ord, symbols))
print(f"\n转换为列表后的结果：{mapped_list}")
print(f"说明：$=36, ¢=162, £=163, ¥=165, €=8364, ¤=164")
````

### 步骤 3：lambda 表达式 - 匿名函数


````python
# lambda 是创建匿名函数的快捷方式
# lambda c: c > 127 等价于以下普通函数：

def greater_than_127(c):
    return c > 127

# 测试 lambda 表达式
test_value = 162
result1 = (lambda c: c > 127)(test_value)
result2 = greater_than_127(test_value)

print(f"lambda c: c > 127 的语法说明：")
print(f"  - lambda: 关键字，表示这是一个匿名函数")
print(f"  - c: 参数名（可以是任何名字，这里表示码点值）")
print(f"  - c > 127: 函数体，返回布尔值")
print(f"\n测试：{test_value} > 127 ?")
print(f"  lambda 结果：{result1}")
print(f"  普通函数结果：{result2}")
````

### 步骤 4：filter() 函数 - 过滤满足条件的值


````python
symbols = '$¢£¥€¤'

# filter(函数, 可迭代对象) 返回一个迭代器，包含所有使函数返回 True 的元素
# filter(lambda c: c > 127, map(ord, symbols))
#    过滤条件     要过滤的数据源

# 先得到所有的码点
all_codes = list(map(ord, symbols))
print(f"所有码点：{all_codes}")

# 然后过滤出大于 127 的
filtered_result = filter(lambda c: c > 127, all_codes)
print(f"\nfilter() 返回的类型：{type(filtered_result)}")
print(f"filter() 对象：{filtered_result}")

# 转换为列表
filtered_list = list(filtered_result)
print(f"过滤后的结果：{filtered_list}")
print(f"说明：只保留了 > 127 的值：162, 163, 165, 8364, 164")
````

### 步骤 5：完整的执行流程演示


````python
symbols = '$¢£¥€¤'

print("=" * 60)
print("完整执行流程演示")
print("=" * 60)

# 原始数据
print(f"\n1. 原始字符串：{symbols}")
print(f"   字符：{list(symbols)}")

# map 阶段
print(f"\n2. map(ord, symbols) - 转换为 Unicode 码点：")
mapped = map(ord, symbols)
mapped_list = list(mapped)
print(f"   结果：{mapped_list}")
print(f"   解释：每个字符 -> 对应的 Unicode 码点")

# filter 阶段
print(f"\n3. filter(lambda c: c > 127, ...) - 过滤大于 127 的值：")
# 注意：map 返回的迭代器已经被消费了，需要重新创建
filtered = filter(lambda c: c > 127, map(ord, symbols))
filtered_list = list(filtered)
print(f"   结果：{filtered_list}")
print(f"   解释：只保留码点 > 127 的值（非 ASCII 字符）")

# 最终结果
print(f"\n4. list(...) - 转换为列表：")
final_result = list(filter(lambda c: c > 127, map(ord, symbols)))
print(f"   最终结果：{final_result}")

# 一行代码的等价写法
print(f"\n5. 一行代码执行：")
beyond_ascii = list(filter(lambda c: c > 127, map(ord, symbols)))
print(f"   beyond_ascii = {beyond_ascii}")
````

### 语法总结

**完整代码：**
```python
beyond_ascii = list(filter(lambda c: c > 127, map(ord, symbols)))
```

**执行顺序（从内到外）：**
1. `symbols` → 字符串 `'$¢£¥€¤'`
2. `map(ord, symbols)` → 将每个字符转换为 Unicode 码点的迭代器：`[36, 162, 163, 165, 8364, 164]`
3. `filter(lambda c: c > 127, ...)` → 过滤出大于 127 的码点：`[162, 163, 165, 8364, 164]`
4. `list(...)` → 将迭代器转换为列表

**关键概念：**
- **map()**: 对可迭代对象的每个元素应用函数，返回迭代器
- **filter()**: 根据函数条件过滤元素，返回迭代器
- **lambda**: 创建匿名函数的快捷语法
- **list()**: 将迭代器转换为列表（因为迭代器是惰性求值的）

**与列表推导式的对比：**
```python
# 函数式风格（当前代码）
beyond_ascii = list(filter(lambda c: c > 127, map(ord, symbols)))

# 列表推导式（更 Pythonic，更易读）
beyond_ascii = [ord(s) for s in symbols if ord(s) > 127]
```
